# Encounter Geometry

Simple geometric proxy for encounter cross-section and detection horizon.

```python
import numpy as np
import matplotlib.pyplot as plt

radius = np.linspace(0.1, 5.0, 100)
cross_section = np.pi * radius**2

plt.plot(radius, cross_section)
plt.title('Encounter Cross-Section vs. Detection Radius')
plt.xlabel('Detection Radius (AU)')
plt.ylabel('Cross-Section (AU^2)')
plt.grid(True)
plt.show()
```
