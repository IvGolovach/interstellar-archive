# Monte Carlo Sketch

Rough simulation for encounter counts over a fixed horizon.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
lambda_rate = 0.3  # events per year
years = 20
trials = 5000
counts = rng.poisson(lambda_rate * years, size=trials)

plt.hist(counts, bins=range(counts.min(), counts.max() + 2), alpha=0.75)
plt.title('Monte Carlo Encounter Counts')
plt.xlabel('Events in 20 Years')
plt.ylabel('Frequency')
plt.grid(axis='y')
plt.show()
```
