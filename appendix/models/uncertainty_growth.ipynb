# Uncertainty Growth Model

Sketch how uncertainty evolves across observation windows.

```python
import numpy as np
import matplotlib.pyplot as plt

windows = np.arange(1, 11)
sigma0 = 1.0
growth = sigma0 * np.sqrt(windows)

plt.plot(windows, growth, marker='o')
plt.title('Uncertainty Growth Across Observation Windows')
plt.xlabel('Observation Window')
plt.ylabel('Relative Uncertainty (σ)')
plt.grid(True)
plt.show()
```
